# Problem 2: Finding a Patient Record

90 minutes | 100 marks | Complete this section only if you selected Problem 2.

A clinic keeps patient IDs in ascending order. A receptionist needs the zero-based index of a requested ID, or -1 if absent. Begin with these distinct IDs:

[102, 118, 125, 139, 147, 163, 180, 204, 219, 235, 248]


## 2.1 Plan and implement the core solution

**Plan.** Name your algorithm and design approach. Explain the search boundaries, your midpoint rule, and the termination condition.


Write your answer here.


**Code.** Use the sorted order to reduce the search; avoid inspecting every record in the worst case. Write a function returning both the index and the number of array positions inspected. Count one inspection whenever an array element is read to compare it with the target.


```python
def binary_search(arr, target):

    low = 0
    high = len(arr) - 1
    inspections = 0
    
    while low <= high:
        mid = low + (high - low) // 2
        inspections += 1
        
        if arr[mid] == target:
            return mid, inspections
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
            
    return -1, inspections

original_ids = [102, 118, 125, 139, 147, 163, 180, 204, 219, 235, 248]
print("Binary Search Test (Target 163):", binary_search(original_ids, 163))
```

## 2.2 Required extension

**Code (efficient search).** The clinic now permits repeated IDs in its test data. Adapt your efficient search to return the FIRST occurrence of a matching ID, or -1. Continue narrowing the search after a match; do not scan left through duplicates. Test it on [102, 118, 118, 118, 139, 163, 163, 180].


```python
def binary_search_first_occurrence(arr, target):

    low = 0
    high = len(arr) - 1
    result_idx = -1
    inspections = 0
    
    while low <= high:
        mid = low + (high - low) // 2
        inspections += 1
        
        if arr[mid] == target:
            result_idx = mid
            high = mid - 1
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
            
    return result_idx, inspections

repeated_ids = [102, 118, 118, 118, 139, 163, 163, 180]
print("Efficient First-Occurrence (118):", binary_search_first_occurrence(repeated_ids, 118))
```

**Code (linear search).** Write a separate linear search that also returns the first occurrence and an inspection count. Use it as a reference to check your efficient implementation and compare the work done. Do not use built-in search functions.


```python
def linear_search_first_occurrence(arr, target):

    inspections = 0
    for i in range(len(arr)):
        inspections += 1
        if arr[i] == target:
            return i, inspections
        elif arr[i] > target:
            break
    return -1, inspections

repeated_ids = [102, 118, 118, 118, 139, 163, 163, 180]
print("Linear Reference First-Occurrence (118):", linear_search_first_occurrence(repeated_ids, 118))
```

## 2.4 Explain and reflect


**Question a.** Explain why discarded positions cannot contain the required answer, including the first-occurrence case.


_Discarded positions cannot contain the target because binary search compares the target with the midpoint value. So, since the array is strictly sorted, if the target is smaller than `arr[mid]`, all elements at the right of `mid` are guaranteed to be larger than the target, making them impossible candidates. The same logic applies reciprocally for the left side. Like for first-occurrence searches, when `arr[mid] == target`, I can safely discard elements at the right because we are searching for an earlier occurrence, also `mid` itself is stored as a potential result before narrowing `high = mid - 1`._


**Question b.** Define n and justify the best-case and worst-case time complexity of each implementation.


_Let $n$ be the number of elements in the list. For efficient binary search, the best-case time complexity is $O(1)$ (when the target is found at the middle on the first check) and the worst-case time complexity is $O(\log n)$ inspections because the search space is halved at each step. For linear search, the best-case is $O(1)$ and the worst-case is $O(n)$ inspections._


**Question c.** Describe how doubling n affects the observed counts and relate this to your complexity claims.


_Doubling $n$ in binary search increases the worst-case inspection count by at most 1 (since $\log_2(2n) = \log_2(n) + 1$), demonstrating logarithmic scaling. In linear search, doubling $n$ approximately doubles the worst-case inspection count, illustrating linear scaling._


**Question d.** Explain what would fail if the list were unsorted.


_If the list were unsorted, the invariant that allows us to discard half the array based on a single comparison (`arr[mid] < target` or `> target`) would break entirely, causing the algorithm to miss valid targets._


**Question e.** Discuss whether detecting unsorted input would change the total complexity. Do not add a full-list sortedness check to the search itself. Distinguish the search cost from any optional validation cost.


_Adding a full-list sortedness check prior to searching would take $O(n)$ time, which would dominate and degrade the overall worst-case complexity of binary search from $O(\log n)$ down to $O(n)$. Therefore, sorted input is assumed, keeping search cost strictly separate from validation._
